## Primeras validaciones y criterios a seguir

**Corrección del encoding**

Gènere aparece como GÃ¨nere, Càrrec como CÃ rrec, Itàlia como ItÃ lia.

Esto es un problema de encoding (UTF-8 leído como Latin-1).

Definimos una función que detecte Latin-1 y cambie a UTF-8 cuando sea necesario, primero corregimos las columnas y luego el resto de los valores de texto.

````python
def fix_encoding(texto):
    """Corrige texto UTF-8 leído como Latin-1, pero solo si hace falta."""
    if not isinstance(texto, str):
        return texto
    try:
        return texto.encode('latin-1').decode('utf-8')
    except (UnicodeEncodeError, UnicodeDecodeError):
        # Si falla, el texto ya estaba bien codificado
        return texto

# Corregimos los nombres de las columnas
encuesta.columns = [fix_encoding(c) for c in encuesta.columns]

# Finalmente corregimos el resto del texto
for col in encuesta.select_dtypes(include='object').columns:
    encuesta[col] = encuesta[col].apply(fix_encoding)
````

**Criterio de decisión para DNIs duplicados**

El DNI es el identificador único de cada persona, y si aparece repetido hay que analizar si la información asociada es coherente antes de decidir qué hacer.

| Situación | Decisión | Justificación |
|---|---|---|
| **Mismos `Nom`, `Cognoms`, `Ciutat`, `Salari`, `Càrrec`** | Duplicado real → conservar 1 | Es la misma persona con los mismos datos repetida. Conservar una copia es suficiente; el resto no aporta información nueva y distorsionaría futuras consultas. |
| **Mismos `Nom` y `Cognoms` pero distinto `Ciutat`/`Salari`/`Càrrec`** | Conflicto → Descartar todos los registros de ese DNI | No hay forma fiable de saber cuál de las versiones es la correcta. Imputar o elegir una arbitrariamente sería fabricar datos. Se descartan todas para no introducir sesgo. |
| **Mismo DNI pero `Nom`/`Cognoms` distintos** | DNI compartido → Descartar todo | Un DNI no puede pertenecer a dos personas distintas. Si los nombres no coinciden, hay un error en el identificador o se está usando el mismo DNI para más de una persona. En ambos casos los registros no son fiables. |

- **No se conserva** un DNI en conflicto: inventar campos como ciudad o el salario "más probable" sería fabricar datos.
- **No se elige** "el que parece más creíble": se descartan **todos** los registros del DNI en conflicto para evitar sesgo de selección.
- **Solo se conserva uno** cuando los registros son idénticos en todos los campos relevantes (duplicado real).

````python
# Eliminar TODOS los registros de DNIs conflictivos
# (los que tras eliminar duplicados exactos siguen teniendo más de una fila)
conteo_dni = encuesta['DNI'].value_counts()
dnis_conflictivos = conteo_dni[conteo_dni > 1].index.tolist()
print("DNIs conflictivos:", dnis_conflictivos)

antes = len(encuesta)
encuesta = encuesta[~encuesta['DNI'].isin(dnis_conflictivos)]
print(f"Filas eliminadas por DNI conflictivo: {antes - len(encuesta)}")
display(encuesta)
````

**¿Dato omitido o dato irreal? - Criterio de eliminación**

En una **encuesta voluntaria de trabajadores**, un campo vacío **no es siempre un error**: puede ser una decisión deliberada de privacidad o un "prefiero no contestar". Por lo tanto, **no se elimina la fila si hay valores nulos**. Lo que sí se revisa, es si los valores **informados** son **reales y/o lógicos**.

| Tipo de problema | ¿Eliminar la fila? | Acción |
|---|---|---|
| **Dato omitido** (DNI, salario, género, edad, ciudad...) | **No** | Se deja como `NaN`. Es información no disponible, no errónea. |
| **Valor imposible** (día `-5` o `32`, mes `0` o `99`, año `3000`, km `-50`) | **No** | Se **anula el valor** (`NaN`). El resto de la fila sigue siendo válido. |
| **Valor ilógico pero posible** (salario `673 €`, km `50`, consumo `0.5`) | **No** | Se **marca como outlier** para revisarlo, pero no se anula sin más. |
| **DNI mal formado** (`XXXX`, `###`) | **No** (si hay Nom/Cognoms) | Se **anula el DNI** (`NaN`). Puede ser anonimato o error de tecleo. |
| **Fila completamente vacía** | **Sí** | No hay ningún dato del que partir. |
| **Filas de prueba** (`???`, `Narnia`, `molts`) | **Sí** | No representa a ningún trabajador real. |
| **DNI en conflicto** (una misma persona con datos repetidos) | **Sí** (todos los registros de ese DNI) | No se puede saber cuál es el correcto; elegir uno sería fabricar datos. |

- **Omisiones**: un nulo por privacidad es válido y la fila se conserva.
- **Se revisa la lógica**: día fuera de 1–31, mes fuera de 1–12, año fuera de rango, km negativos, salarios absurdos.
- **Ante un valor imposible, se anula la celda, no la fila**: se pierde menos información anulando un valor, que eliminando toda una fila.
- **Cuándo elimino la fila** cuando no queda ningún dato útil, o cuando el registro no representa a un trabajador real.

# Normalización de valores categóricos y numéricos

## Columna Càrrec

Primero damos un vistazo a todos los valores de la columna para luego tomar decisiones.



````python
encuesta['Càrrec'].value_counts(dropna=False).sort_index()
````

### Criterio de normalización para Càrrec

a) Analista/Analyst → unificar en una sola forma

Cap de projecte/Project lead → unificar en una sola forma

Es la misma palabra (catalán/inglés).

Conservamos Analista y Cap de projecte que están en catalán.

b) Analista/Analista junior → NO unificar

Senior analyst/Analista → NO unificar

Son cargos distintos.

Unificarlos destruiría información (nivel de seniority).

c) Senior analyst está en Inglés → Traducir al catalán para mejorar la uniformidad de la encuesta.

Data analyst → Se deja en inglés porque es un término muy asentado en el sector.

Tomadas las decisiones, hacemos un **mapa de normalización**.

````python
import numpy as np

mapa_carrec = {
    # Variantes de mayúscula/abreviatura → forma canónica
    'data analyst':    'Data Analyst',
    'analyst':         'Analista',
    'admin.':          'Administratiu',
    'cap de projecte': 'Cap de projecte',

    # Mismo cargo con otro nombre (traducción) → forma canónica
    'project lead':    'Cap de projecte',

    # Traducción al catalán para concordancia
    'senior analyst':  'Analista sènior',
}

def normalitza_carrec(v):
    if pd.isna(v):
        return np.nan
    return mapa_carrec.get(str(v).strip().lower(), str(v).strip())

encuesta['Càrrec'] = encuesta['Càrrec'].apply(normalitza_carrec)
encuesta['Càrrec'].value_counts(dropna=False).sort_index()
````

## Columna Gènere

Al igual que con Càrrec, damos un pantallazo previo a la toma de decisiones.

````python
encuesta['Gènere'].value_counts(dropna=False)
````

### Criterios de normalización para Gènere

- **Variantes en la escritura → unificar `D` / `Dona` / `F` estas son la misma categoría con distintos valores (inicial, abreviatura, género). Lo mismo para `H` / `Home` / `M`.
- **`M` → `Home`.** Se interpreta como inicial de *Masculí* en el contexto catalán del dataset, consistente con `D` (*Dona*) y `H` (*Home*), que también son iniciales. Se documenta esta decisión porque en castellano `M` también puede significar *Mujer*; se adopta el criterio catalán por coherencia con el resto de códigos de la columna.
- **`NC` → `NaN`.** No contesta, lo tratamos como a sin dato. La persona no aporta información de género, por lo cual, no se crea una categoría. Se unifica al resto de valores ausentes (`NaN`).
- **`Altre`** se mantiene como categoría propia. Es una **respuesta real** distinta de Dona/Home, no una omisión.
- **Clave del mapa en minúsculas + `.strip()`.** Una sola entrada (`'d'`) captura `D`, `d`, `D `, etc., sin duplicar entradas por mayúsculas o espacios.

### Nota sobre posibles inconsistencias nombre–género

> En una revisión cruzada entre `Gènere` y `Nom` pueden aparecer casos donde el nombre no coincide con el género declarado. **No se corrige ni se anula automáticamente**, porque:

> - La columna `Gènere` refleja cómo la persona **se identifica**, no cómo la perciben los demás. Una persona puede identificarse como *Dona* y tener un nombre tradicionalmente masculino (personas trans, no binarias, o simplemente nombres culturalmente ambiguos).
> - **Asumir que el nombre define el género** introduciría un sesgo y podría discriminar a personas LGTBI+. El dato de género **declarado por la propia persona es la fuente de verdad**, no el nombre.
> - Corregir `Dona` → `Home` (o viceversa) por el nombre sería **fabricar el dato**, no limpiarlo.

> Por tanto, se respeta el género declarado en la encuesta y el cruce con `Nom` se usa **solo** como pista para detectar códigos ambiguos (como fue el caso de `M`), no para reasignar géneros individuales.

````python
mapa_genere = {
    # → Dona
    'd':    'Dona',
    'dona': 'Dona',
    'f':    'Dona',

    # → Home
    'h':    'Home',
    'home': 'Home',
    'm':    'Home',

    # → Altre
    'a':    'Altre',

    # → Sin dato
    'nc':   np.nan,
}

def normalitza_genere(v):
    if pd.isna(v):
        return np.nan
    return mapa_genere.get(str(v).strip().lower(), str(v).strip())

encuesta['Gènere'] = encuesta['Gènere'].apply(normalitza_genere)

encuesta['Gènere'].value_counts(dropna=False)
````

## Criterio de normalización para Salari Mensual

Los textos no numéricos encontrados son **solo estos cuatro**

| Texto | Significado | Destino |
|---|---|---|
| `n/d` | No disponible | `NaN` |
| `mil vuit-cents` | 1800 | `1800` |
| `dos mil` | 2000 | `2000` |
| `tres mil` | 3000 | `3000` |

Las acciones a tomar serían las siguientes

| Formato | Ejemplo | Acción |
|---|---|---|
| Coma | `1,469 €` | eliminar `,` → `1469` |
| Punto| `1.021 €` | eliminar `.` → `1021` |
| Símbolo `€` | `1479€` | eliminar |
| Texto `euros` | `1358 euros` | eliminar |
| Texto catalán | `mil vuit-cents`, `dos mil`, `tres mil` | traducir a número |
| `n/d` | `n/d` | `NaN` |
| Cualquier otro no-dígito | espacios, letras | eliminar con `re.sub` |

> Punto y coma son AMBOS separadores de miles. No hay decimales en el salario, así que se eliminan directamente. Esto evita la ambigüedad típica de "¿`12,5` es doce coma cinco o ciento veinticinco?".

> Quitar todo lo que tenga relacionado a euro para dar uniformidad a los valores en las filas, pero a su vez dejar la unidad en el nombre de la columna renombrando a Salari mensual (€) para no perderla.

> No hay un criterio de máximos y mínimos ya que al ser una encuesta no se sabe si es su salario real, un trabajador autónomo que solo hace hrs en la empresa, o cualquier otro tipo de caso.



````python
import re

textos_cat = {
    'n/d':              np.nan,
    'mil vuit-cents':   1800,
    'dos mil':          2000,
    'tres mil':         3000,
}

def parse_salari(v):
    if pd.isna(v):
        return np.nan
    if isinstance(v, (int, float)):
        return float(v)

    s = str(v).strip().lower()

    # Textos no numéricos (incluye n/d → NaN)
    if s in textos_cat:
        return textos_cat[s]

    # Quitar "euros", "€" y espacios
    s = s.replace('euros', '').replace('€', '').strip()

    # Coma y punto son AMBOS separadores de miles → eliminar
    s = s.replace('.', '').replace(',', '')

    # Quitar cualquier cosa que no sea dígito
    s = re.sub(r'[^\d]', '', s)

    if s == '':
        return np.nan

    try:
        return float(s)
    except ValueError:
        return np.nan

encuesta['Salari mensual'] = encuesta['Salari mensual'].apply(parse_salari)

# Cambio de nombre de la columna
encuesta = encuesta.rename(columns={'Salari mensual': 'Salari mensual (€)'})

display(encuesta["Salari mensual (€)"].value_counts(dropna=False))
````

## Criterio de unificación y normalización para Fills / No Fills / Nombre_fills

De las tres columnas conservaré **dos**:

- **`Fills`** → booleano (tiene hijos sí/no).
- **`Nombre_fills`** → número de hijos (entero nullable, que admite `NaN`).

Se **elimina** `No Fills`, porque su información queda derivada de `Fills` (`Fills = False` ⟺ no tiene hijos). Mantener ambas sería redundante.

| Situación | `Fills`| `Nombre_fills` | Motivo |
|---|---|---|---|
| `Nombre_fills` > 0 | `True` | ese número | Tener un número > 0 implica tener hijos |
| `Nombre_fills` == 0 | `False` | `0` | 0 hijos = no tiene hijos |
| `Nombre_fills` es `NaN` (y sin más info) | `False` | `NaN` | Sin número ni marca → se asume "no" |
| `Fills == True` sin número | `True` | `NaN` | Tiene hijos pero no sabemos cuántos → no inventamos `1` |
| `No Fills == True` sin número | `False` | `NaN` (o `0`) | No tiene hijos |
| `Nombre_fills` con valor **y** `No Fills == True` | `NaN` | `NaN` | **Incongruencia**: dice tener hijos pero marca "no tiene" → todo a `NaN` |
| `Fills == True` **y** `No Fills == True` | `NaN` | `NaN` | **Incongruencia**: ambas marcas contradictorias → todo a `NaN` |
| Todo vacío / `NaN` | `False` | `NaN` | Sin dato; `Fills=False` por defecto |

### Criterio de incongruencia

Ante cualquier **contradicción** entre las columnas (un número de hijos junto a `No Fills = True`, o `Fills` y `No Fills` ambas `True`), se anula **todo** (`Fills → NaN`, `Nombre_fills → NaN`). No se elige una versión sobre otra porque no hay forma fiable de saber cuál es la correcta, y elegir arbitrariamente introduciría sesgo.

### Criterio final del enunciado

El enunciado pide quedarnos solo con una columna de las tres. Por lo tanto, me quedo con `Nombre_fills` que es la que reune la información más completa de todas.

````python
def unifica_fills(row):
    n        = row['Nombre_fills']
    fills    = row['Fills']
    no_fills = row['No Fills']

    # ---------- Incongruencias → NaN ----------

    # 1) Número de hijos junto a No Fills == True
    if pd.notna(n) and n > 0 and no_fills == True:
        return pd.Series({'Fills': np.nan, 'Nombre_fills': np.nan})

    # 2) Fills y No Fills ambas True
    if fills == True and no_fills == True:
        return pd.Series({'Fills': np.nan, 'Nombre_fills': np.nan})

    # ---------- Casos coherentes ----------

    # Con número declarado
    if pd.notna(n):
        n_int = int(n)
        if n_int > 0:
            return pd.Series({'Fills': True, 'Nombre_fills': n_int})
        return pd.Series({'Fills': False, 'Nombre_fills': 0})

    # Sin número: usar Fills / No Fills
    if fills == True:
        return pd.Series({'Fills': True, 'Nombre_fills': np.nan})
    if no_fills == True:
        return pd.Series({'Fills': False, 'Nombre_fills': np.nan})

    # Sin dato
    return pd.Series({'Fills': False, 'Nombre_fills': np.nan})

# Aplicar y reconstruir las dos columnas
resultat = encuesta.apply(unifica_fills, axis=1)
encuesta['Fills']        = resultat['Fills'].astype('boolean')
encuesta['Nombre_fills'] = resultat['Nombre_fills'].astype('Int64')

# Eliminar No Fills
encuesta = encuesta.drop(columns=['Fills', 'No Fills'])
````

## Análisis con porcentajes y ratios

El enunciado pide reflexionar sobre si las medias absolutas permiten una comparación justa cuando los grupos tienen tamaños distintos.

La respuesta es no: una media sobre 2 personas no es comparable a una media sobre 200.

Los porcentajes y ratios corrigen ese problema.

### Tamaño de cada grupo (para contextualizar)

````python
print("\nPersonas por género × país:")
display(
    encuesta
    .groupby(['Gènere', "País d'origen"])
    .size()
    .unstack(fill_value=0)
)
````

### Ratio salarial respecto a la media global

En lugar de la media absoluta, se puede expresar cada grupo como ratio respecto a la media global (1.0 = igual que la media; 1.2 = 20% por encima; 0.8 = 20% por debajo).

````python
mitjana_global = encuesta['Salari mensual (€)'].mean()

ratio_genere = (
    encuesta
    .groupby('Gènere')['Salari mensual (€)']
    .mean()
    .div(mitjana_global)
    .sort_values(ascending=False)
    .round(2)
    .rename('Ratio vs mitjana global')
)

display(ratio_genere)
````

### Ratio género × país respecto a la media del país

Esto es más informativo: compara cada género dentro de su país respecto a la media de ese país. Así se elimina el efecto de que un país tenga salarios más altos en general.

```
  1.0 = mismo salario que la media de su país.

> 1.0 = por encima de la media del país.

< 1.0 = por debajo.
```

````python
# Media por país
mitjana_pais = encuesta.groupby("País d'origen")['Salari mensual (€)'].transform('mean')

# Ratio de cada persona respecto a la media de su país
encuesta['ratio_vs_pais'] = encuesta['Salari mensual (€)'] / mitjana_pais

ratio_creuat = (
    encuesta
    .pivot_table(
        index='Gènere',
        columns="País d'origen",
        values='ratio_vs_pais',
        aggfunc='mean',
        margins=True,
        margins_name='Mitjana total'
    )
    .round(2)
)

display(ratio_creuat)
````

### Porcentaje de personas por encima de la mediana global, por grupo

Otro enfoque basado en porcentajes (no en medias) en donde cada celda es el % de personas de ese grupo que cobran por encima de la mediana global.

Es una medida robusta y comparable entre grupos de distinto tamaño, porque es un porcentaje, no un valor absoluto.

````python
mediana_global = encuesta['Salari mensual (€)'].median()

pct_sobre_mediana = (
    encuesta
    .assign(sobre_mediana=encuesta['Salari mensual (€)'] > mediana_global)
    .groupby(['Gènere', "País d'origen"])
    ['sobre_mediana']
    .mean()
    .mul(100)
    .round(2)
    .unstack()
)

display(pct_sobre_mediana)
````

### Comparación: media absoluta vs ratio

````python
comparacio = pd.DataFrame({
    'Mitjana (€)': encuesta.groupby('Gènere')['Salari mensual (€)'].mean(),
    'N': encuesta.groupby('Gènere')['Salari mensual (€)'].count(),
    'Ratio vs global': ratio_genere,
}).round(2)

display(comparacio)
````

### Medias absolutas: útiles pero engañosas con grupos desiguales

La tabla de salario medio por género y país muestra diferencias de cientos de euros entre grupos. Sin embargo, **las medias absolutas no permiten una comparación justa cuando los tamaños de los grupos son muy distintos**

- Un grupo con 2 personas puede tener una media muy alta o muy baja por azar, sin que eso refleje una tendencia real.
- Un país con 200 empleados y otro con 5 no son comparables en términos absolutos: la media del país pequeño tiene mucha más varianza.

### Por qué los porcentajes y ratios son más objetivos

- **Ratio vs media global / vs media del país:** normaliza cada grupo respecto a una referencia, eliminando el efecto de que unos países tengan salarios más altos en general. Un ratio de `1.2` significa "20% por encima de la referencia", independientemente del tamaño del grupo.
- **% de personas por encima de la mediana global:** es una **proporción**, no una media. Es robusta a outliers y directamente comparable entre grupos de distinto tamaño. Un grupo con `70%` de personas por encima de la mediana está claramente mejor pagado que uno con `30%`, sin importar cuántas personas haya en cada uno.

### Cómo cambia la interpretación

- Con **medias absolutas**, un país con pocos empleados muy bien pagados puede parecer "el mejor" cuando en realidad es un artefacto de muestra pequeña.
- Con **ratios**, ese mismo país puede quedar en una posición más modesta si se compara con su propia referencia.
- Con **porcentajes**, se ve la **distribución** dentro del grupo, no solo el promedio: un grupo puede tener una media alta pero mucha desigualdad interna (pocos muy altos y muchos bajos), y eso el porcentaje lo revela mejor.

### Conclusión

Las medias absolutas sirven para una **primera visión**, pero para **comparar grupos de tamaños distintos** conviene acompañarlas de:
1. El **tamaño del grupo** (`N`).
2. **Ratios** respecto a una referencia (global o del país).
3. **Porcentajes** (proporción por encima de la mediana, distribución por tramos).
